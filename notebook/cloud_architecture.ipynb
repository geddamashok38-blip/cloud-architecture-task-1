# Cloud Architecture Design — Task 1

Provider-neutral architecture for a hypothetical cloud-based business management application.

## 1. Architecture Components

- DNS
- CDN / Load Balancer
- Private application compute
- Managed relational database
- Object storage
- Cache
- Monitoring/logging
- Backup and disaster recovery


## 2. Network Boundaries

Internet-facing services are isolated from private application and database tiers. The database has no public endpoint. Firewall/security-group rules permit only required communication.

## 3. Data Flow

User → DNS → HTTPS edge/load balancer → application instance → database/object storage/cache → response.

## 4. Availability

Application compute spans at least two availability zones. Failed instances are removed by health checks and replaced by auto-scaling. Managed database HA is preferred where supported.

## 5. Backup and DR

Assumed RPO: 15 minutes. Assumed RTO: 1 hour. Automated database backups, object versioning, separate backup storage, and restore testing are used.

## 6. Scalability

Horizontal application scaling is the primary approach. Database optimization/read replicas and independent object storage scaling support future growth.
